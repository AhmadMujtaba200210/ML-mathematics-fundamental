### Computing Derivative with Differentiation

$y = x^2 + 2x +2$

```python
def f(my_x):
    my_y = my_x**2 + 2*my_x + 2
    return my_y
```

```python
f(2)
```

```python
y = f(x)
```

```python
fig,ax = plt.subplots()
plt.axvline(x=0,color='lightgray')
plt.axhline(y=0,color='lightgray')
_=plt.plot(x,y)
```

lets identify where is the slop, when x=2

```python
x1=2
y1=f(x1)
y1
```

```python
fig,ax = plt.subplots()
plt.axvline(x=0,color='lightgray')
plt.axhline(y=0,color='lightgray')
plt.scatter(2,10)
_ = plt.plot(x,y)
```

##### the delta method uses the difference between two points to calculate the slop. To illustrate this, lets define another point, Q where, say, x = 5

```python
f(5)
```

```python
fig, ax = plt.subplots()
plt.axvline(x=0,color='lightgray')
plt.axhline(y=0,color='lightgray')
plt.scatter(2,10)
plt.scatter(5,37,color='orange',zorder=3)
_ = ax.plot(x,y)
```

To find the slop m between points P and Q:
$$m=\frac{\text{change in y}}{\text{change in x}} = \frac{\Delta y}{\Delta x} = \frac {y_2-y_1}{x_2-x_1}=\frac{37-10}{5-2}=\frac{27}{3}=9$$

```python
m = (37-10)/(5-2)
m
```

To plot the line passes through P and Q, we can rearrange the equation of a line y = mx + b solve for b:
$$b=y - mx$$

```python
b = 37-m*5
b
```

```python
line_y = m*x + b
```

```python
fig,ax = plt.subplots()
plt.axvline(x=0,color='lightgray')
plt.axhline(y=0,color='lightgray')
plt.scatter(2,10)
plt.scatter(5,37,color='orange',zorder=3)
plt.ylim(-5,150)
plt.plot(x, line_y, c='orange')
_ = ax.plot(x,y)
```

The closer Q becomes to P, the closer the slope m comes to the true tangent of the point P. Lets demonstrate this with another point Q

Previously, our $\Delta x$ between Q and P was equal to 3. Now it is much smaller 
$$\Delta x = x_2-x_1=2.1-2=0.1$$

```python
x2=2.1
y2=f(x2)
y2
```

```python
fig,ax = plt.subplots()
plt.axvline(x=0,color='lightgray')
plt.axhline(y=0,color='lightgray')
plt.scatter(2,10)
plt.scatter(2.1,10.61,zorder=3)
_ = ax.plot(x,y)
```

##### Calculate the slope y, As according to the formula of slope: y = b + mx
So we will need to find out first m then b m
so for m it should be like, $\frac{\Delta y}{\Delta x}\text{ where it is equal to,}\frac{y_2-y_1}{x_2-x_1} $

```python
m = (y2-y1)/(x2-x1)
m
```

```python
#now b where, b = y-mx
b = 10.61-m*2.1
b
```

```python
line_y = m*x + b
```

```python
fig,ax = plt.subplots()
plt.axvline(x=0,color='lightgray')
plt.axhline(y=0,color='lightgray')
plt.scatter(2,10)
plt.scatter(2.1,10.67,zorder=3)
plt.ylim(-5,150)
plt.plot(x,y)
_=ax.plot(x,line_y)
```

$\text{As Q becomes infinitely close to P:}$
- $x_2 - x_1$ approaches 0
- In other words, $\Delta {x} \text{ approaches 0}$
- This can be denoted by $\Delta {x} \to 0$

```python
def diff_demo(my_f,my_x,my_delta):
    return (my_f(my_x+my_delta)-my_f(my_x))/my_delta
```

```python
deltas = [1,0.1,0.01,0.001,0.0001,0.00001,0.000001]
```

```python
for delta in deltas:
    print(diff_demo(f,2,delta))
```

```python
for delta in deltas:
    print(diff_demo(f,-1,delta))
```

#### Regression in PyTorch
From here, we will use the pytoch automatic differentiation library to fit a straight line into data points.

```python
import torch
import matplotlib.pyplot as plt
```

```python
x = torch.tensor([0,1,2,3,4,5,6,7.]) # E.g. Dosage of drugs to treat Alzheimer's treatment
x
```

The y values will be created using line equation i.e. y = mx + b
so, lets say we have a slope y of m = -0.5 and b=2

```python
y = -0.5*x + 2 + torch.normal(mean=torch.zeros(8),std=0.2)
y
```

```python
fig, ax = plt.subplots()
plt.title('Clinal Trial')
plt.ylabel('Forgetfulness')
plt.xlabel('Drug Dosage (ml)')
plt.plot(x,y)
_ = ax.scatter(x,y)
```

Initialize m to any random value from above plane

```python
m = torch.tensor([0.9]).requires_grad_()
m
```

Do the same for b 

```python
b = torch.tensor([0.1]).requires_grad_()
b
```

```python
def regression(my_x,my_m,my_b):
    return my_x*my_m + my_b
```

```python
def regression_plot(my_x, my_y, my_m, my_b):
    fig, ax = plt.subplots()
    ax.scatter(my_x, my_y)

    x_min, x_max = ax.get_xlim()

    # Safely detach scalar values from the computation graph
    my_m = my_m.detach().item()
    my_b = my_b.detach().item()

    # y_min, y_max = m_val * x_min + b_val, m_val * x_max + b_val
    y_min = regression(x_min,my_m,my_b)
    y_max = regression(x_max,my_m,my_b)

    ax.set_xlim([x_min, x_max])
    _ = ax.plot([x_min, x_max], [y_min, y_max], color='red')
```

```python
regression_plot(x,y,m,b)
```

As we have seen line has not fitted properly to our datapoints, so we will see how to fit this regression line to out datapoints

### Machine Learning

#### Step 1: Forward Pass

```python
yhat = regression(x,m,b)
yhat
```

#### Step 2: Compare $\hat y$ with y to calculate the Cost
There is a pytorch Loss method MSELoss method, but lets define it ourselves, to see how it works. MSE Cost is defined by:
$$C = \frac{1}{n} \sum_{i=2}^{n} (\hat y_i-y_i)^2$$

```python
def mse(my_yhat, my_y):
    sigma = torch.sum((my_yhat - my_y)**2)
    return sigma/len(my_y)
```

```python
C = mse(yhat,y)
C
```

#### Step 3: Use Auto Diff to calculate gradient of C w.r.t parameters

```python
C.backward()
```

```python
m.grad
```

```python
b.grad
```

The one with bigger value either it is m or b will have the more affect on this model, here you can see the m has more greater value, so its affect will be much bigger

#### Step 4: Gradient Descent

```python
optimizer = torch.optim.SGD([m,b],lr=0.01) # lr - learning rate: this tells how much we have adjust 
```

```python
optimizer.step()
```

Now much better but still needed to be fit along the datapoints.  
So, we will repeat those step once again 
but lets wait, why we don't create a method similar to Epoch, we have seen many times when finetunning any LLM or any AI model

```python
epochs = 1000
for epoch in range(epochs):
    optimizer.zero_grad() # Reset the gradients to zero, else they accumulate

    yhat = regression(x,m,b) # Step 1: Forward Pass
    C = mse(yhat,y) # Step 2: Calculating Cost

    C.backward() # Step 3: Automatic Differentiation (Cacluate Gradient)
    optimizer.step() # Step 4: optimize  it (Calculate Gradient Descent)

    print(f'Epoch {epoch}, cost {C.item()}, m grad {m.grad.item()}, b grad {b.grad.item()}')
```

```python
regression_plot(x,y,m,b)
```
