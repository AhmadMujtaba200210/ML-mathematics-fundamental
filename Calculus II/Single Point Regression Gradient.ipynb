```python
import torch
```

In this notebook, we calculate the gradient of quadratic cost with respect to a straight-line regression model's parameters.

```python
xs = torch.tensor([0,1,2,3,4,5,6,7]) #no of milimeters for alzimer drug given to patient
```

```python
ys = torch.tensor([1.86, 1.31, .62, .33, .09, -.67, -1.23, -1.37]) #outcome of treatment by drug i.e. forgetfullness is decreased as dosage increased
```

Slope of Line given by: $y = mx +b$

```python
def regression(x,m,b):
    return m*x+b
```

```python
m = torch.tensor([0.9]).requires_grad_()
```

```python
b = torch.tensor([0.1]).requires_grad_()
```

To keep the partial derivatives as simple as possible, let's move forward with a single instance $i$ from the eight possible data points: 

```python
i = 7
x = xs[i]
y = ys[i]
```

```python
x
```

```python
y
```

**Step 1**: Forward pass

```python
yhat = regression(x,m,b)
yhat
```

**Step 2**: Compare $\hat{y}$ with true $y$ to calculate cost $C$

In the *Regression in PyTorch* notebook, we used mean-squared error, which averages quadratic cost over multiple data points. With a single data point, here we can use quadratic cost alone. It is defined by: $$ C = (\hat{y} - y)^2 $$

```python
def mean_squared_error(yhat,y):
    return torch.square(yhat-y)
```

```python
C = mean_squared_error(yhat,y)
C
```

**Step 3**: Use autodiff to calculate gradient of $C$ w.r.t. parameters

```python
C.backward()
```

```python
m.grad # partial derivative w.r.t m
```

```python
b.grad # partial derivative w.r.t b
```
