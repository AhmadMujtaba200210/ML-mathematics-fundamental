```python
from scipy.integrate import quad # "quadrature" = numerical integration (as opposed to symbolic)
```

 $$ \int_1^2 \frac{x}{2} dx = \frac{3}{4} $$

```python
def g(x):
    return x/2
```

```python
quad(g,1,2) # here the 1 and 2 are the limits typed in function above as integration range
```

The second output of quad is an estimate of absolute error of the integral, which is in this case is absolute zero

###### Another example:
$$\int_3^4 2xdx = \frac{3}{4}$$

```python
def h(x):
    return x*2
```

```python
quad(h,3,4)
```
